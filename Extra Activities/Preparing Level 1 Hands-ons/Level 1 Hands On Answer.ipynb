# Task 1 : Scalar × Vector

## Example :

**Input :** \
\
scalar = 3 \
vector = [2, 4, 5]


**Output :** \
[6, 12, 15]


## Code :

```python
def task_1(scalar , vector):

  """
  Multiply a scalar value by every element in a vector and return the resulting vector.
  """

  result = [scalar * x for x in vector]

  return result
```

```python
print("Running Task 1: Scalar × Vector : ")
scalar = 3
vector = [2, 4, 5]
print("Input : ")
print(f"scalar : {scalar}")
print(f"vector : {vector}")
print("Output:", task_1(scalar, vector))
```

# Task 2 : Matrix × Matrix Multiplication

## Example :

**Input :** \
\
A = [ \
  [1, 2], \
  [3, 4] \
] \
\
B = [ \
  [2, 0], \
  [1, 2] \
]



**Output :** \
[ \
  [4, 4], \
  [10, 8] \
]



## Code :

```python
def task_2(A , B):
  """
  Multiply two matrices A (m×n) and B (n×p) and return the resulting matrix.
  """

  # Number of rows in A
  m = len(A)
  # shared dimension
  n = len(A[0])
  # columns in B
  p = len(B[0])

  # Create result matrix filled with zeros (m × p)
  result = [[0 for _ in range(p)] for _ in range(m)]

   # Perform matrix multiplication
  for i in range(m):           # rows of A
      for j in range(p):       # columns of B
          for k in range(n):   # shared dimension
              result[i][j] += A[i][k] * B[k][j]


  return result
```

```python
print("Running Task 2: Matrix × Matrix Multiplication : ")
A = [
    [1, 2],
    [3, 4]
]
B = [
    [2, 0],
    [1, 2]
]
print("Matrix A:", A)
print("Matrix B:", B)
print("Output:", task_2(A, B))
```

# Task 3 : Check if Matrix is Symmetric

## Example

**Input :** \
\
A = [ \
  [1, 2, 3], \
  [2, 5, 7], \
  [3, 7, 9] \
]




**Output :** \
True

## Code

```python
def task_3(A):
  """
  Check if a matrix is symmetric.
  A matrix is symmetric if : A = Transpose of A
  """
  rows = len(A)
  cols = len(A[0])

  # Must be square
  if rows != cols:
    return False

  # Check symmetry: A[i][j] == A[j][i]
  for i in range(rows):
      for j in range(cols):
          if A[i][j] != A[j][i]:
              return False

  return True
```

```python
print("Running Task 3: Check if Matrix is Symmetric : ")
M = [
    [1, 2, 3],
    [2, 5, 7],
    [3, 7, 9]
]
print("Matrix:", M)
print("Output:", task_3(M))
```
