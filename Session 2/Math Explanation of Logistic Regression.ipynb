# **Cost function**


```python
import numpy as np
e = np.e

def cost_function(x, y, w, b):
  """
  Computes the cost function for logistic regression.

  Args:
    x[m][n] : Data, m examples, n number of features
    y[m]    : target values
    w[n]    : weights of the model
    b       : bias of the model

  Returns:
    total_cost : The cost of using w,b as the parameters for logistic regression
  """
  """
  we need to change the range of the function from [-infinity : +infinity] to [0 : 1]
  so we can not use the simple linear regression equation (wx + b)
  we will use the sigmoid function:  1 / (1 + e**(-z)) where z = wx + b
  then we will use the logistic regression cost function:
  J(w,b) = - (1 / m) * sum( y * log(h(x)) + (1-y) * log(1-h(x)) ) for i = 1 to m
  where h(x) is the sigmoid function
  
  -> for part : y * log(h(x)) : if y = 1 -> we want h(x) to be 1 (correct prediction) so the log(1) = 0 (no cost / no error)
                                if y = 0 -> we don't care about this part (because it will be 0) so the log(0) -> -infinity (high cost / high error)
                                
  -> for part : (1-y) * log(1-h(x)) : if y = 0 -> we want h(x) to be 0 (correct prediction) so the log(1-0) = log(1) = 0 (no cost / no error)
                                      if y = 1 -> we don't care about this part (because it will be 0) so the log(1-1) = log(0) -> -infinity (high cost / high error)
  thus the cost function will penalize wrong predictions heavily and reward correct predictions with low cost.
  
  this process implemented for each training example and then averaged over all training examples to get the final cost.
  """
  # number of training examples
  m = x.shape[0]

  total_cost = 0

  #TODO: compute the cost function result

  # prediction = 1 / (1 + e**(-(np.dot(x, w) + b)))  # Sigmoid function
  # total_cost = - (1 / m) * np.sum(y * np.log(prediction) + (1 - y) * np.log(1 - prediction))
  for i in range(m):
    z = np.dot(x[i] , w) + b # -> wx + b (linear equation)
    sigmoid = 1 / (1 + e ** (-z))  # Sigmoid function
    total_cost += - (y[i] * np.log(sigmoid) + (1 - y[i]) * np.log(1 - sigmoid)) # -> y - log(h(x)) - (1-y)log(1-h(x))
  total_cost = total_cost / -m

  #END OF CODE

  return total_cost
```

# **Gradient/Derivative terms**


```python
def compute_partial_derivatives(x, y, w, b):
  """
  Computes the partial derivatives required for gradient descent of logistic regression
  Args:
    x[m][n] : Data, m examples, n number of features
    y[m]    : target values
    w[n]    : weights of the model
    b       : bias of the model

  Returns:
    dj_dw[n] : The gradient of the cost w.r.t. the parameters w[n]
    dj_db    : The gradient of the cost w.r.t. the parameter b
  """
  # Number of training examples
  m, n = x.shape

  dj_dw = np.zeros(n)
  dj_db = 0

  #TODO: compute the gradients

  for i in range(m):
    sigmoid = 1/(1 + e ** (-np.dot(x[i], w)))
    error = sigmoid - y[i]

    for j in range(n):
      dj_dw[j] += error * x[i][j]

    dj_db += error

  dj_dw /= m
  dj_db /= m
  

  #END OF CODE

  return dj_dw, dj_db
```

# **Gradient Descent**


```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters):
  """
  Performs gradient descent to fit w,b. Updates w,b by taking
  num_iters gradient steps with learning rate alpha

  Args:
    x[m][n]       : Data, m examples, n number of features
    y[m]          : target values
    w_in[n], b_in : initial values of model parameters
    alpha         : Learning rate
    num_iters     : number of iterations to run gradient descent

  Returns:
    w[n] : Updated values of parameters after running gradient descent
    b    : Updated value of parameter after running gradient descent
  """
  # Number of training examples
  m, n = x.shape

  w = w_in.copy()
  b = b_in

  #TODO: update the weights using gradient descent algorithm
  for i in range(num_iters):
    dj_dw, dj_db = compute_partial_derivatives(x, y, w, b)
    w -= alpha * dj_dw
    b -= alpha * dj_db


  #END OF CODE

  return w, b #return w and b
```
